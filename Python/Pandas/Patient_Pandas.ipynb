```python
#🔹 SECTION 2: FILTERING & CONDITIONAL SELECTION (16–35)
#16.	Find all male patients
df[df["Gender"] == "Male"]



#17.	Find all female patients above 50 years
df[(df["Gender"] == "Female") & (df["Age"] > 50)]

#18.	Find patients from Pune city
df[df["City"] == "Pune"]

#19.	Find patients suffering from Diabetes
df[df["Chronic_Condition"] == "Diabetes"]

#20.	Find patients who are smokers
df[df["Smoker"] == "Yes"]

#21.	Find patients who consume alcohol occasionally
df[df["Alcohol_Consumption"] == "Sometimes"]

#22.	Find patients with BMI greater than 30
df[df["BMI"] > 30]

#23.	Find patients with no chronic condition
df[df["Chronic_Condition"] == "None"]

#24.	Find patients registered after 2023
df[df["Registration_Date"] > "2023-01-01"]

#25.	Find patients with blood group O+
df[df["Blood_Group"] == "O+"]

#26.	Find patients insured under Star Health
df[df["Insurance_Plan"] == "Star Health"]

#27.	Find patients whose height is above 175 cm
df[df["Height"] > 175]

#28.	Find patients aged between 30 and 45
df[(df["Age"] > 30) & (df["Age"] < 45)]

#29.	Find patients from Mumbai OR Pune
df[(df["City"] == "Mumbai") | (df["City"] == "Pune")]

#30.	Find patients who are married and diabetic
df[(df["Marital_Status"] == "Married") & (df["Chronic_Condition"] == "Diabetes")]

#31.	Find patients whose email contains “gmail”
df[df["Email"].str.contains("gmail")]

#32.	Find patients whose Patient_ID starts with “P1”
df[df["Patient_ID"].str.startswith("P1")]

#33.	Find patients with weight above average weight
df[df["Weight"] > df["Weight"].mean()]

#34.	Find patients having Asthma or Thyroid
df[(df['Chronic_Condition'] == 'Asthma') | (df['Chronic_Condition'] == 'Thyroid')]

#35.	Find patients whose BMI is below 18.5
df[df["BMI"] < 18.5]

#36 Remove age-gruop column
#df.drop(columns = "age_group", inplace = True)

# Usages if loc and iloc function
df.loc[4,['Last_Name', "Age"]]
```

```python
#56.	Create a new column Age_Group (Young, Adult, Senior)

df["Age_Group"] = df["Age"].apply(lambda i :  "Senior" if i  > 60 else ("Adult" if i > 18 else "Young"))
```

```python
#73.	Sort patients by BMI descending
df.head(2)

df.sort_values("BMI", ascending  = False)

#74.	Reset index after sorting
df.reset_index(drop = True)

#76.	Find top 5 cities with highest average BMI
df.groupby("City")["BMI"].mean().sort_values(ascending= False).head()
```
