```python
#25. Create heatmap of correlation matrix (Age, BMI, Height, Weight)


# 1. Calculate the correlation matrix for specific columns
# Use .corr() to find relationships between -1 (negative) and +1 (positive)
corr_matrix = df[['Age', 'BMI', 'Height_cm', 'Weight_kg']].corr()

# 2. Set up the matplotlib figure
plt.figure(figsize=(10, 8))

# 3. Create the heatmap
sns.heatmap(corr_matrix,
            annot=True,       # Displays the correlation value in each cell
            cmap='coolwarm',  # Color palette (blue for negative, red for positive)
            fmt=".2f",        # Rounds values to 2 decimal places
            linewidths=0.5,   # Adds small gaps between cells
            vmin=-1, vmax=1)  # Ensures the color scale is fixed from -1 to 1

plt.title('Correlation Heatmap: Age, BMI, Height, & Weight', fontsize=16)
plt.show()
```

```python
#26. Create pivot-based heatmap of City vs Chronic_Condition
# 1. Create a pivot table counting occurrences of each condition per city
# We use .size() to count and .unstack() to format it for a heatmap
pivot_data = df.groupby(['City', 'Chronic_Condition']).size().unstack(fill_value=0)

# 2. Plot the heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(pivot_data,
            annot=True,     # Show the actual count in each cell
            fmt='d',        # Use integer format for the counts
            cmap='YlGnBu',  # Yellow-Green-Blue color scale
            linewidths=.5)  # Add small gaps between cells

plt.title('Prevalence of Chronic Conditions by City', fontsize=16)
plt.xlabel('Chronic Condition', fontsize=12)
plt.ylabel('City', fontsize=12)
plt.show()
```

```python
#28. Create facet grid of BMI distribution by Gende
# 1. Initialize the FacetGrid
# 'col' creates a separate subplot for each Gender
grid = sns.FacetGrid(df, col="Gender", height=4, aspect=1.2, hue="Gender")

# 2. Map a distribution plot (histogram + KDE) to the grid
grid.map(sns.histplot, "BMI", kde=True, bins=20, alpha=0.6)

# 3. Add titles and labels
grid.set_axis_labels("BMI ($kg/m^2$)", "Frequency")
grid.set_titles(col_template="{col_name} BMI Distribution")

plt.show()
```
