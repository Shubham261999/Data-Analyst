```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/Python/Pandas/employee_data.csv")
df.head(1)
```

```python
#1.	Create a bar chart showing the number of employees in each department.
# using group by stored this result in varibale
a= df.groupby("Department")["First Name"].count()
plt.figure(figsize=(10,5))
a.plot(kind = "bar", color = "green")
plt.title("Number of employees in each department")
plt.xlabel("Department")
plt.ylabel("Number of Employee")
plt.show()
```
